For even dimensions we contruct table of index *pairs* over *slots*.
There is one slot for each +1 in the involution template.
 

```python
import clifford.inverse.model_inverse.py as Ref
import clifford.context as Clif

Cl(10) # so Clif.Grade[] table is as big as we need.
print(len(Clif.Grade))

dims = (4,6,8,10)
invos = (Ref.invo4, Ref.invo6, Ref.invo8, Ref.invo10)

def make_slot_list( template ):
    slot_list = []
    for n in range(len(template)):
        if 1 == template[n]:
            slot_list += [n]
    return slot_list

slot_lists = [make_slot_list(template) for template in invos]

def make_slot_pairs(slot_list):
    ex_count = 0
    slot_pairs = [[] for _ in range(len(slot_list))]
    for s1 in slot_list:
        for s2 in slot_list:
            s12 = s1^s2
            if s12 in slot_list:
                slot_pairs[s12] += [(s1,s2)]
            else:
                ex_count += 1
    for n in range(len(slot_pairs)):
        print(n,len(slot_pairs[n]))
    if ex_count:
        print("less {ex_count}")
    return slot_pairs

slot_pair_lists = []
for n in range(len(dims)):
    print(f"\ndimension = dims[n]")
    slot_pair_lists += [make_slot_pairs(slot_lists[n])]                    
        
                    
                    
```
