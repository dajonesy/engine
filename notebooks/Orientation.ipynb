A vector orientation is tradtional.  A spacial orientation is equally tenable.

When we add another dimension to a model, we usually think of including another orthogonal direction appended to the frame of directions we already model.  It would be equally valid to extend the *space* to which we already confine ourselves.  These spaces are just the first pseudoscalars added to the model as dimensions are accreted.  In the ordinal layout, the first elements of each grade can be distinquished, and recruited as an alternative basis.  This notebook is a study of this alternative orientation.



```python
import numpy as np
import clif.tools as tools
import clif.util as util

dimensions = 7
bfmt = f"0{dimensions}b"
dfmt = f"{len(str(1<<dimensions))}d"
spacial_basis = [(1<<n)-1 for n in range(1,dimensions+1)]
elements = tools.sub_algebra( spacial_basis )

spacial_sign_table = tools.sign_table( elements )
spacial_row_keys = np.empty( 1<<dimensions, dtype=int )
for r in range(len(elements)):
    key_index = tools.sign_row_to_index( spacial_sign_table[r] )
    spacial_row_keys[r] = key_index
#    row_string = tools.sign_row_to_string( spacial_sign_table[r] )
#    print(f"{r:{dfmt}} {r:{bfmt}}  {elements[r]:{dfmt}} {elements[r]:{bfmt}}  {key_index:2d} {key_index:{bfmt}}  {row_string}")
```

```python
print( util.xor_test( spacial_row_keys ) )
spacial_row_key_basis = [spacial_row_keys[1<<n] for n in range(dimensions)]
for n in range(dimensions):
    m = 1<<n
    print(f"{m:{bfmt}} {elements[m]:{bfmt}} {spacial_row_keys[m]:{bfmt}}")
```

Notice the pattern in the bit-places over the elements (column 4).
Bit_0: `0110100110010110`.  `0 1 10 1001 10010110`
The pattern goes by the name "Thue-Morse".
The pattern is established by iteration; the size doubling each iteration.
Extend the pattern by attaching the pattern's complement.
The bit_1 pattern is similar, only we begin with `00` instead of `0`.
In general, begin with $2^n$ zeros for bit_n.

```python
transposed_spacial_sign_table = tools.transpose( spacial_sign_table )
spacial_col_keys = np.empty( 1<<dimensions, dtype=int )
for r in range(len(elements)):
    col_key_index = tools.sign_row_to_index( transposed_spacial_sign_table[r] )
    spacial_col_keys[r] = col_key_index
#    col_string = tools.sign_row_to_string( transposed_spacial_sign_table[r] )
#    print(f"{r:2d} {r:04b} {elements[r]:2d} {elements[r]:04b} {col_key_index:2d} {col_key_index:04b} {col_string}")
```

The pattern presented by the column keys seems more organized than the row keys.
- Each odd column key iterates the prior even column key
- Each odd iterate is the complement of the prior even interate
After establishing this much of the pattern it is difficult to characterize further.


```python
print( util.xor_test( spacial_col_keys ) )
#spacial_col_key_basis = [spacial_col_keys[1<<n] for n in range(dimensions)]
for n in range(dimensions):
    m = 1<<n
    xor_keys = spacial_row_keys[m] ^ spacial_col_keys[m]
    print(f"{n:{dfmt}} {m:{bfmt}} {elements[m]:{bfmt}} {spacial_row_keys[m]:{bfmt}} {spacial_col_keys[m]:{bfmt}}"
         + f"  {xor_keys:{bfmt}}")
```

A pattern in the spacial key basis (row or column) is not discernable.
However, there is a pattern that can be described in the xor of the two bases.

- The odd xor bases are built from '01's
- The even bases are built by reversing the bits in the odd base, and reversing the order.
- `srk[1]==sck[2]`, `srk[3]==sck[4]`, `srk[5]==sck[6]`
- `srk[1]==sck[1]<<1`, `srk[3]==sck[3]<<1`, `srk[5]==sck[5]<<1`, `srk[7]==sck[7]<<1`
